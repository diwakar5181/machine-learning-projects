```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import scipy.cluster.hierarchy as sc
from sklearn.cluster import AgglomerativeClustering
```

### Objectives
1. Load and explore the dataset for clustering.
2. Prepare and standardize the feature data for hierarchical clustering.
3. Use principal component analysis (PCA) to reduce the feature space for visualization.
4. Build and interpret a dendrogram to help select an appropriate number of clusters.
5. Train an agglomerative clustering model and visualize the resulting clusters.

```python
# let's load iris dataset
dataset = load_iris()
print(dataset.DESCR)
```

```python
dataset.keys()
```

```python
df = pd.DataFrame(dataset.data, columns=dataset.feature_names)
df.head()
```

```python
# let's scale our independent features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
# apply PCA and reduce number of features
pca = PCA(n_components=2)
pca_scaled = pca.fit_transform(X_scaled)
```

```python
# visualize if pca dimensionality reduction did a good enough job of capturing variance in the data 
plt.scatter(pca_scaled[:,0], pca_scaled[:, 1], c=dataset.target)
```

```python
# Experiment with Agglomerative clustering by constructing a dendogram
plt.figure(figsize=(8,6))
plt.title('Dendogram')

sc.dendrogram(sc.linkage(pca_scaled, method='ward'))
plt.title('Dendogram')
plt.xlabel('Sample Index')
plt.ylabel('Eucledian Distance')
```

```python
# create agglomerative clustering model

model = AgglomerativeClustering(n_clusters=2, linkage='ward')
model.fit(pca_scaled)
```

```python
model.labels_
```

```python
# visualize clustering
plt.scatter(pca_scaled[:,0], pca_scaled[:, 1], c=model.labels_)
```
