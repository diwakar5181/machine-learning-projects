```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

### Objectives:-
1. Load breast_cancer data from sklearn
2. Create data frame
3. Perform standardization using StandardScaler()
4. Apply PCA and reduce number of features to 2 and the ones which captures the maximum variance in data
5. Plot 2 features with the dependent feature as hue (color) to understand if it did a good job of clustering and separating
    the predictable values

```python
dataset = load_breast_cancer()
```

```python
print(dataset.DESCR)
```

```python
dataset.keys()
```

```python
df = pd.DataFrame(data=dataset['data'], columns=dataset['feature_names'])
```

```python
df.head()
```

```python
# perform standardization

scaler = StandardScaler()
scaled_df = scaler.fit_transform(df)
```

```python
# Apply PCA Algorithms to reduce number of columns to 2. PCA will reduce columns to 2 columns that captures the highest
# variance in the dataset

pca = PCA(n_components=2)
pca_df = pca.fit_transform(scaled_df)
```

```python
pca_df
```

```python
# let's look at the pca calculated variance of the two columns it captured
pca.explained_variance_
```

```python
# now let's visualize how pca did when comes it comes to clustering the data for predicting the 
# dependent feature (Malignant/Benign)

plt.figure(figsize=(8,6))
plt.scatter(pca_df[:, 0], pca_df[:, 1], c=dataset['target'], cmap='plasma')
plt.xlabel('First Principle Component')
plt.ylabel('Second Principle Component')
```
