```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# create dataset for a classification problem
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=42)
```

```python
# visualize independent features
pd.DataFrame(X)
```

```python
# Visualize dependent feature
pd.DataFrame(y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
# Model training
from sklearn.linear_model import LogisticRegression

logistic_reg = LogisticRegression()
```

```python
logistic_reg.fit(X_train, y_train)
```

```python
y_pred = logistic_reg.predict(X_test)
print(y_pred)
```

```python
# Let's look into how model did predictions or calculated the probabilities for X_test
logistic_reg.predict_proba(X_test) 
```

```python
# performance metrics
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
score = accuracy_score(y_test, y_pred)
print(score)

cm = confusion_matrix(y_test, y_pred)
print(cm)

cr = classification_report(y_test, y_pred)
print(cr)
```

## Hyperparameter Tuning And Cross Validation

```python
log_model = LogisticRegression()
params = {'penalty': ['l1','l2', 'elasticnet'], 'C': [100, 10, 1, 0.1, 0.01], 'solver': ['lbfgs', 'liblinear', 'newton-cg', 'sag', 'saga']}
```

```python
# GridSearchCV
# It will help find best parameter fit for the model for a specific dataset

from sklearn.model_selection import GridSearchCV, StratifiedKFold, RandomizedSearchCV

cv = StratifiedKFold()
grid = GridSearchCV(estimator=log_model, param_grid=params, scoring='accuracy', cv=cv)
```

```python
grid.fit(X_train, y_train)
```

```python
# find out which params gave model the best accuracy/score 
print(grid.best_params_)
print(grid.best_score_)
```

```python
# let's do predictions and see how model performed
y_pred = grid.predict(X_test)

score = accuracy_score(y_test, y_pred)
print(score)
cm = confusion_matrix(y_test, y_pred)
print(cm)
cr = classification_report(y_test, y_pred)
print(cr)
```

### As you can see our scores have increased with fewer False-Positives and False-negatives

```python
# Randomized Search CV will randomly pick params for best fit line or model accurasy

random_cv = RandomizedSearchCV(estimator=log_model, param_distributions=params, scoring='accuracy', cv=5)
random_cv.fit(X_train, y_train)
```

```python
print(random_cv.best_score_)
print(random_cv.best_params_)
```

```python
y_pred = random_cv.predict(X_test)

score = accuracy_score(y_test, y_pred)
print(score)
cm = confusion_matrix(y_test, y_pred)
print(cm)
cr = classification_report(y_test, y_pred)
print(cr)
```

## Logistic Regression for Imbalanced Dataset

```python
# create an imbalanced dataset with param "weights=""
X, y = make_classification(n_samples=1000, n_features=2, n_clusters_per_class=1, random_state=10, n_redundant=0, weights=[0.99])
```

```python
# count binary values to show imbalance 
pd.DataFrame(y).value_counts()
```

```python
# visualizing imbalanced dataset with teh help of sns
sns.scatterplot(x=pd.DataFrame(X)[0], y=pd.DataFrame(X)[1], hue=y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python
# getting a combination of class weights 
imb_dataset_model = LogisticRegression()
class_weights = [{0:w, 1:y} for w in [1,10,50,100] for y in [1,10,50,100]]
params = {'penalty': ['l1','l2', 'elasticnet'], 'C': [100, 10, 1, 0.1, 0.01], 'solver': ['lbfgs', 'liblinear', 'newton-cg', 'sag', 'saga'], 'class_weight':class_weights}
class_weights
```

```python
cv = StratifiedKFold()
grid = GridSearchCV(estimator=imb_dataset_model, param_grid=params, scoring='accuracy', cv=cv)
grid.fit(X_train, y_train)
```

```python
# '1' classifier was given 5x importance over '0' classifier
grid.best_params_
```

```python
y_pred = grid.predict(X_test)

score = accuracy_score(y_test, y_pred)
print(score)
cm = confusion_matrix(y_test, y_pred)
print(cm)
cr = classification_report(y_test, y_pred)
print(cr)
```
