```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

### Objectives:-
1.  Implement decision tree classifier
2. Calculate metrics with default model params
3. Visualize decision tree with default params to understand purity/impurity 
4. Post-prune tree and rerun the model
5. Compare result with the original model
6. Pre-pruning or hyperparameter tuning with GridSearchCV
7. Find best params

```python
iris = load_iris()
```

```python
print(iris.DESCR)
```

```python
X, y = pd.DataFrame(iris.data, columns=iris.feature_names), pd.DataFrame(iris.target)
print(X)
print(y)
```

```python
# train, test, split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

```python
# apply decision tree classifier with default params
model_classifier = DecisionTreeClassifier()
model_classifier.fit(X_train, y_train)
```

```python
# let's visualize decision tree
plt.figure(figsize=(15,7))
tree.plot_tree(model_classifier, filled=True)
```

### As you can see it has used Gini coefficient to calculate purity/impurity
1. Left split (leaf)has reached purity with value (40, 0, 0) so no further splits
2. It will continue and split right node until it reaches the purity which is again calculated by Gini Coefficient
3. This will lead to overfitting so we will look into post-pruning and pre-pruning

```python
# let's run predictions

y_pred = model_classifier.predict(X_test)

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
# let's try with post-pruning and prune the tree to second level
model_post_pruine = DecisionTreeClassifier(max_depth=2)
model_post_pruine.fit(X_train, y_train)
```

```python
# let's visualize pruned decision tree
plt.figure(figsize=(15,7))
tree.plot_tree(model_post_pruine, filled=True)
```

```python
# let's run predictions with pruned model

y_pred = model_post_pruine.predict(X_test)

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
# Hyperparameter tuning for pre-pruning

params = {
    "criterion" : ["gini", "entropy", "log_loss"],
    "splitter": ["best", "random"],
    "max_depth": [1,2,3,4,5],
    "max_features":["auto", "sqrt", "log2"]
          }
```

```python
import warnings

warnings.filterwarnings('ignore')
grid = GridSearchCV(DecisionTreeClassifier(), param_grid=params, scoring='accuracy', cv=5)
grid.fit(X_train, y_train)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

```python
y_pred = grid.predict(X_test)

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```
