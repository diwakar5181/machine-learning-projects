```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, roc_curve
```

```python
# Generate 2 class dataset
X,y = make_classification(n_samples=1000, n_classes=2, random_state=1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python
# create a dummy model with default output as 0
dummy_model_prob = [0 for _ in range(len(y_test))]
dummy_model_prob
```

```python
# create a basic Logistic model
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
# get models predicted probabilities - we will focus on positive side only
model_prob = model.predict_proba(X_test)

model_prob = model_prob[:, 1]
```

```python
dummy_model_auc = roc_auc_score(y_test, dummy_model_prob)
model_auc = roc_auc_score(y_test, model_prob)

print(dummy_model_auc)
print(model_auc)
```

```python
# calculate FPR, TPR and threshold for both dummy and model with psotive probabilities

dummy_fpr, dummy_tpr, dummy_threshold = roc_curve(y_test, dummy_model_prob)
model_fpr, model_tpr, model_threshold = roc_curve(y_test, model_prob)
```

```python
model_threshold
```

```python
# plot roc curve for both the models

plt.plot(dummy_fpr, dummy_tpr, linestyle='--', label='Dummy Model')
plt.plot(model_fpr, model_tpr, marker='.', label='Logistic Model')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')

plt.legend()
plt.show()
```

### Conclusion:- We can clearly see and as expected, the area under the curve of the Logistic model with positive probabilities is greater than the area under curve of the Dummy Model with 0 probabilities. Which means that Logistic Model has performed better than dummy model on PR (Positive Rate) 

```python
# Let's plot threshold for the Logistic Model on the same figure and visualize on how to determine the best possible threshold.
# For example as you get cloer to a 1 threshold you will get higher TPR but your FPR will increase as well
fig = plt.figure(figsize=(10,30))
plt.plot(dummy_fpr, dummy_tpr, linestyle='--', label='Dummy Model')
plt.plot(model_fpr, model_tpr, marker='.', label='Logistic Model')
ax = plt.subplot(1,1,1)
for xyz in zip(model_fpr,model_tpr,model_threshold):
    ax.annotate("%s" % np.round(xyz[2],2), xy=(xyz[0],xyz[1]))

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')

plt.legend()
plt.show()
```
